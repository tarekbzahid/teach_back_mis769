```python
import assemblyai as aai
import os

# Try to import from config.py, fallback to environment variable
try:
    from config import ASSEMBLYAI_API_KEY, AUDIO_FILE_PATH
except ImportError:
    print("Warning: config.py not found. Using environment variables or defaults.")
    print("Please copy config.example.py to config.py and add your API key.")
    ASSEMBLYAI_API_KEY = os.getenv("ASSEMBLYAI_API_KEY", "your_api_key_here")
    AUDIO_FILE_PATH = "Custom-Home-Builder.mp3"

# Set up the API key
aai.settings.api_key = ASSEMBLYAI_API_KEY
transcriber = aai.Transcriber()

# Use relative path to audio file
transcript = transcriber.transcribe(AUDIO_FILE_PATH)
```

```python
# show the transcripted text from the call record file
text= transcript.text
print(text)
```

```python
# Summarizing the transcripted text
# Import the pipeline class from the transformers module

from transformers import pipeline

# Try to import summarization settings from config
try:
    from config import MAX_SUMMARY_LENGTH, MIN_SUMMARY_LENGTH
except ImportError:
    MAX_SUMMARY_LENGTH = 1000
    MIN_SUMMARY_LENGTH = 30

# Load the summarization pipeline
summarizer = pipeline("summarization")

# Transcript from audio transcription
transcript = text

# Generate summary
summary = summarizer(
    transcript, 
    max_length=MAX_SUMMARY_LENGTH, 
    min_length=MIN_SUMMARY_LENGTH, 
    do_sample=False
)[0]['summary_text']

# Print out the summary
print("Summary:")
print(summary.strip())
```
